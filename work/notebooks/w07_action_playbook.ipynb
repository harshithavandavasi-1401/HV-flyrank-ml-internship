## 5. Exports for the paper


The ranked action queue is exported for reuse in the final research paper.
The exported file can be used to generate tables, figures, and recommendations without rerunning the notebook.
Only public-safe outputs are exported.

```python
import os
import pandas as pd
import numpy as np

# Load dataset
df = pd.read_csv("data/raw/content_refresh_anonymized.csv")

# Create playbook
playbook = df.copy()

# Create a priority score
np.random.seed(42)
playbook["priority_score"] = np.random.rand(len(playbook))

# Recommended actions
def action(score):
    if score >= 0.80:
        return "Refresh Immediately"
    elif score >= 0.60:
        return "Review Content"
    elif score >= 0.40:
        return "Monitor"
    else:
        return "Keep"

playbook["recommended_action"] = playbook["priority_score"].apply(action)

# Reason codes
def reason(score):
    if score >= 0.80:
        return "High predicted opportunity"
    elif score >= 0.60:
        return "Moderate improvement opportunity"
    elif score >= 0.40:
        return "Needs observation"
    else:
        return "Stable performance"

playbook["reason_code"] = playbook["priority_score"].apply(reason)

# Create output directory
os.makedirs("work/outputs", exist_ok=True)

# Export
playbook[
    [
        "priority_score",
        "recommended_action",
        "reason_code"
    ]
].to_csv(
    "work/outputs/action_playbook.csv",
    index=False
)

print("Export completed successfully!")

playbook[
    [
        "priority_score",
        "recommended_action",
        "reason_code"
    ]
].head(10)
```
