### My lane as an ML task (type)

**Task Type:** Binary Classification

My lane is **Content Decline Prediction**, which is a binary classification problem. The model predicts whether a content page is likely to experience a significant decline in search impressions during the next evaluation period. The output has two classes: **Declining** and **Not Declining**. This helps content teams identify pages that need attention before search traffic drops.

```python
import pandas as pd

df = pd.read_csv("data/raw/content_refresh_anonymized.csv")

print("Dataset Shape:", df.shape)
print("Number of Columns:", len(df.columns))
```

### Target or proxy

The target is whether a content page experiences a significant decline in search impressions during the prediction period. The label is created from observed Search Console performance by comparing impressions across two time windows. Pages with a decline greater than the chosen threshold are labeled **Declining**, while the remaining pages are labeled **Not Declining**.

```python
print("Target column:", "trend_direction")
print("\nTarget distribution:")
print(df["trend_direction"].value_counts())
```

### Success metric

The primary evaluation metric is **F1-score** because the dataset may contain more non-declining pages than declining pages. A good model should achieve an F1-score above **0.80** while maintaining a balance between precision and recall.

```python
print("Missing values in dataset:")
print(df.isnull().sum().sort_values(ascending=False).head(10))
```

### The unit of analysis

Each row in the dataset represents a single content page (URL) observed over a specific time period. The row contains features such as impressions, clicks, CTR, average position, and other historical performance metrics that are used to predict whether that page will experience a future decline in search impressions.

```python
print(df[[
    "search_volume",
    "impressions_90d",
    "clicks_90d",
    "ctr",
    "avg_position",
    "trend_direction"
]].head())
```

### Why ML beats a fixed rule here

Content performance depends on many interacting factors such as historical impressions, clicks, CTR, average position, seasonal trends, and user behavior. A simple rule like "CTR below 3%" or "impressions dropped by 20%" cannot capture these complex relationships. A machine learning model can learn patterns from multiple features simultaneously and provide more accurate predictions.

```python
numeric_features = [
    "search_volume",
    "impressions_90d",
    "clicks_90d",
    "sessions_90d",
    "ctr",
    "avg_position",
    "content_age_days",
    "trend_pct"
]

df[numeric_features].describe()
```
